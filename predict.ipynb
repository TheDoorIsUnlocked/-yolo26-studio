```python
from ultralytics import YOLO
yolo = YOLO(model='icon-V11.pt',task='detect')
result = yolo(source='001.png',save=True,conf=0.66,save_txt=True,show=True)
```

```python
display(result[0].names)
display(result[0].boxes.cls)
display(result[0].boxes.xywh)
```

```python
def get_icon_id(icon_name:str):
    for i in result[0].names:
        if result[0].names[i] == icon_name:
            return i
def get_cls_no(icon_id:int):
    for idx,item in enumerate(result[0].boxes.cls):
        if item == icon_id:
            return idx  
def get_xy(cls_no:int):
    x = result[0].boxes.xywh[cls_no][0]
    y = result[0].boxes.xywh[cls_no][1]
    return x,y
pos = get_xy(0)
display(pos)
```
